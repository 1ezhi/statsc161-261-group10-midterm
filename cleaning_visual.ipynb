# Data cleaning funnel
From 677k markets to 35k usable markets

```python
# Funnel chart: from 677k markets to 35k usable markets
import matplotlib
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.family": ["Liberation Sans", "DejaVu Sans"], "font.size": 12,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.edgecolor": "#9b9a96", "axes.labelcolor": "#52514e",
    "xtick.color": "#52514e", "ytick.color": "#52514e",
    "axes.titleweight": "bold", "axes.titlesize": 15, "axes.titlelocation": "left",
    "axes.grid": True, "grid.color": "#e6e4df", "axes.axisbelow": True,
})
BLUE, GREY, INK, INK2 = "#2a78d6", "#9b9a96", "#0b0b0b", "#52514e"

stages = ["Markets that traded at least once",
          "Active: ≥ $20k volume and ≥ 200 trades",
          "Final: complete history + 12h ahead,\nnot closed or closing within 12h"]
vals = [677390, 39748, 35413]

fig, ax = plt.subplots(figsize=(10, 4.6))
ax.barh(stages[::-1], vals[::-1], color=[GREY, BLUE, BLUE], height=0.6)
for i, v in enumerate(vals[::-1]):
    ax.text(v + 8000, i, f"{v:,}", va="center", fontsize=13, color=INK, fontweight="bold")
ax.text(35413 + 8000, -0.33, "→ 1.34M prediction moments (every 4 hours)",
        fontsize=11, color=INK2, va="center")
ax.set_xlim(0, 820000)
ax.xaxis.set_major_formatter(matplotlib.ticker.FuncFormatter(lambda x, _: f"{x/1000:.0f}k"))
ax.set_title("From 677k markets to 35k usable markets")
ax.grid(axis="y", visible=False)
ax.set_xlabel("number of markets")
fig.tight_layout(rect=(0, 0.04, 1, 1))
fig.text(0.01, 0.01, "Source: Polymarket trading data, all markets, May 30 to Oct 25, 2026.",
         fontsize=8.5, color=INK2)
fig.savefig("cleaning_funnel.png", dpi=200)
plt.show()
```
